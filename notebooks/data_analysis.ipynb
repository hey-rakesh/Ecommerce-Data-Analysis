```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv(r"D:\Jupiter\E-commerce_project\data\Ecommerce_clean.csv")
# print(df.head(5))
```

# Overall Business Performance

```python
total_Sales = df["sales"].sum()
total_profit = df["profit"].sum()
total_order = df["order_id"].count()
unique_customer = df["customer_name"].nunique()
aov = (total_Sales / total_order)
margin = (total_Sales / total_profit)
```

# Category Performance Analysis

```python
category_summary = df.groupby("category").agg(
    Total_Sales=("sales", "sum"),
    Total_Profit=("profit", "sum"),
    Total_Quantity_Sold=("quantity", "sum"),
    Average_Discount=("discount_percent", "mean")
)
category_summary["Profit_Margin_%"] = (
    category_summary["Total_Profit"] / category_summary["Total_Sales"]
) * 100

# print(category_summary)

rank_1 = category_summary["Total_Profit"].rank(ascending=True, method="dense")
# print(rank_1)


category_summary["Sales_Rank"] = category_summary["Total_Sales"].rank(ascending=False)

category_summary["Margin_Rank"] = category_summary["Profit_Margin_%"].rank(ascending=True)

# print(category_summary[["Margin_Rank", "Sales_Rank"]])
```

# Product Performance Analysis

```python
top_product_sales = df.groupby("product_name")["sales"].sum().sort_values(ascending=False).head(10)
# print(top_product_sales) 

top_product_profit = df.groupby("product_name")["profit"].sum().sort_values(ascending=False).head(10)
# print(top_product_profit)

product_summary = df.groupby("product_name").agg(
    total_sales = ("sales", "sum"),
    total_profit = ("profit", "sum")
)
# print(product_summary)

high_sales = product_summary["total_sales"].median()
low_profit = product_summary["total_profit"].median()

result = product_summary[(product_summary["total_sales"] > high_sales) & (product_summary["total_profit"] < low_profit)]
# print(result)
```

# Geographic Analysis 

```python
high_state_sales = df.groupby("state")["sales"].sum().sort_values(ascending=False).head(1)
high_state_profit = df.groupby("state")["profit"].sum().sort_values(ascending=False).head(1)

city_summary = df.groupby("city").agg(
    total_sales = ("sales", "sum"),
    total_profit = ("profit", "sum"),
    total_order = ("order_id", "count")
)
city_rank = city_summary[["total_order", "total_sales", "total_profit"]].rank(ascending=False).astype(int)
# print(city_rank)

high_sales = city_summary["total_sales"].median()
low_profit = city_summary["total_profit"].median()

city_result = city_summary[(city_summary["total_sales"] > high_sales) & (city_summary["total_profit"] < low_profit)]
# print(city_result)
```

# Customer Analysis

```python
top_10_customer = df.groupby("customer_id")["sales"].sum().sort_values(ascending=False).head(10)
top_10_profit = df.groupby("customer_id")["profit"].sum().sort_values(ascending=False).head(10)

customer_summary = df.groupby("customer_id").agg(
    total_sales = ("sales", "sum"),
    total_profit = ("profit", "sum"),
    total_orders = ("order_id", "count"),
    avg_order = ("sales", "mean")

)
# print(customer_summary)

high_spending = customer_summary["total_sales"].median()
high_value_costomer = customer_summary[customer_summary["total_sales"] > high_spending]
# print(high_value_costomer)
```

# Discount Impact Analysis

```python
discount_summary = df.groupby("discount_percent").agg(
    total_sales = ("sales", "sum"),
    total_profit = ("profit", "sum")
)
discount_summary["profit_margin"] = (discount_summary["total_profit"] / discount_summary["total_sales"]) * 100

best_profile = discount_summary["total_profit"].max()

best_discount = discount_summary[discount_summary["total_profit"] == best_profile]
# print(best_discount)


```

# Payment Method Analysis

```python
payment_summary = df.groupby("payment_method").agg(
    total_order = ("order_id", "count"),
    total_sales = ("sales", "sum"),
    total_profit = ("profit", "sum"),
    avg_order = ("sales", "mean")
)
# print(payment_summary)

best_profile = payment_summary["total_profit"].max()
best_payment_method = payment_summary[payment_summary["total_profit"] == best_profile]
# print(best_payment_method)
```

# Order Status Analysis

```python
order_summary = df.groupby("order_status").agg(
    total_sales = ("sales", "sum"),
    total_order = ("order_id", "count"),
    total_profit = ("profit", "sum")
)

total_order = df["order_id"].count()
cancelled_orders = df[df["order_status"] == "Cancelled"]["order_id"].count()
cancel_rate = (cancelled_orders / total_order) * 100
# print(cancel_rate)

return_order = df[df["order_status"] == "Returned"]["order_id"].count()
return_rate = (return_order / total_order) * 100
# print(return_rate)
```

# Time-Based Analysis

```python

df["year_month_sales"] = df["order_date"].dt.to_period("M")
monthly_sales = df.groupby("year_month_sales")["sales"].sum()

df["year_month_profit"] = df["order_date"].dt.to_period("M")
monthly_profit = df.groupby("year_month_profit")["profit"].sum()
# print(monthly_profit)

monthly_profit = df.groupby("year_month")["profit"].sum()

best_month_profit = monthly_profit[monthly_profit == monthly_profit.max()]
# print(best_month_profit)

worst_month_profit = monthly_profit[monthly_profit == monthly_profit.min()]
# print(worst_month_profit)

monthly_comparison = df.groupby("year_month").agg(
    total_sales=("sales", "sum"),
    total_profit=("profit", "sum")
)

# print(monthly_comparison)
```
